```python
# Problem 3
print('Problem 3')

def sigrej(data, limits, mask=None):
    '''
    Perform iterative sigma rejection on a data array.

    Parameters
    ----------
    data : numpy array
        Array containing the data to be sigma clipped.

    limits : tuple
        Tuple containing the number of standard deviations to use
        as the rejection limit for each iteration.

    mask : Boolean numpy array, optional
        Boolean mask indicating which data points are initially good.
        True = good, False = bad.
        If no mask is provided, all data points are considered good.

    Returns
    -------
    mask : Boolean numpy array
        Modified mask after sigma rejection.
        True = good, False = rejected.
    '''

    # If no mask was provided, start with all points marked as good
    if mask is None:
        mask = np.ones(data.shape, dtype=bool)
    else:
        mask = mask.copy()

    # Perform sigma rejection for each limit
    for limit in limits:

        # Using good data points
        good_data = data[mask]

        # Calculating mean and standard deviation
        mean = np.mean(good_data)
        sigma = np.std(good_data)

        # Masking out "bad points" (far from mean)
        mask[np.abs(data - mean) > limit * sigma] = False

    return mask



# Running it on data set from question 2
new_mask = sigrej(sample, (5., 5.))

clean_sample = sample[new_mask]

print("Mean of cleaned data:", np.mean(clean_sample))
```
